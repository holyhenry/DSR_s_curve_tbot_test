```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 50
state = np.array([0., 0.])
leader_state = np.array([5.0, 1.])

xval = np.linspace(state[0], leader_state[0], N)
yval = np.interp(xval, [state[0],leader_state[0]], [state[1],leader_state[1]])
```

```python
print(np.array([xval, yval]))
print(np.array([xval, yval]).shape)
print((np.array([xval, yval]).T).tolist())
print(np.array([xval, yval]).T)
print(np.array([xval, yval]).T.shape)
```

```python
plt.plot(np.array([state[0],leader_state[0]]), np.array([state[1],leader_state[1]]),'o')
plt.plot(xval, yval,'.')
```
